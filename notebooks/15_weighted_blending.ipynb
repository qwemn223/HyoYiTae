```python
import pandas as pd
import numpy as np
import os

# 1. 기존에 생성된 가장 성능이 좋은 모델들의 제출 파일 로드
# (파일 이름이 다를 경우 실제 저장된 파일명으로 수정 필요)
sub4 = pd.read_csv('../submissions/submit_04_hyperparameter_finetuning.csv')
sub10 = pd.read_csv('../submissions/submit_10_lgbm_seed_ensemble.csv') 
sub7 = pd.read_csv('../submissions/submit_07_final_stacking.csv')

# 2. 가중 평균 블렌딩 적용
# 최고 성능인 4번(MAE 0.1692)에 50%, 10번(0.1764)에 30%, 7번(0.1802)에 20% 가중치 부여
weight_4 = 0.5
weight_10 = 0.3
weight_7 = 0.2

blended_preds = (sub4['stress_score'] * weight_4) + \
                (sub10['stress_score'] * weight_10) + \
                (sub7['stress_score'] * weight_7)

# 3. 새로운 제출 파일 생성
sub_blend = sub4.copy()
sub_blend['stress_score'] = np.clip(blended_preds, 0, 1)

# 4. 저장
os.makedirs('../submissions', exist_ok=True)
submit_path = '../submissions/submit_12_weighted_blending.csv'
sub_blend.to_csv(submit_path, index=False)
print(f"★ 가중 평균 블렌딩(5:3:2) 제출 파일 생성 완료: {submit_path}")
```
